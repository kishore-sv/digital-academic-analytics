```python
import pandas as pd

df = pd.read_csv("student_data/student-mat.csv", sep=";")

print("Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
display(df.head())
```

```python
academic_columns = [
    "absences",
    "G1",
    "G2",
    "G3",
    "failures",
    "studytime"
]

print("Academic indicator summary:")
display(df[academic_columns].describe())

print("\nUnique values:")
for column in academic_columns:
    print(f"\n{column}:")
    print(sorted(df[column].unique()))
```

```python
df["performance_trend"] = df["G2"] - df["G1"]

print("Performance trend summary:")
display(df["performance_trend"].describe())

print("\nTrend distribution:")
print(
    df["performance_trend"]
    .apply(
        lambda x: "Improving" if x > 0
        else "Stable" if x == 0
        else "Declining"
    )
    .value_counts()
)
```

```python
def absence_risk(absences):
    if absences <= 4:
        return 0
    elif absences <= 9:
        return 1
    else:
        return 2

df["absence_risk"] = df["absences"].apply(absence_risk)

print("Absence risk distribution:")
print(
    df["absence_risk"]
    .value_counts()
    .sort_index()
    .rename({
        0: "LOW",
        1: "MEDIUM",
        2: "HIGH"
    })
)
```

```python
def performance_risk(g2):
    if g2 >= 12:
        return 0
    elif g2 >= 10:
        return 1
    else:
        return 2

df["performance_risk"] = df["G2"].apply(performance_risk)

print("Performance risk distribution:")
print(
    df["performance_risk"]
    .value_counts()
    .sort_index()
    .rename({
        0: "LOW",
        1: "MEDIUM",
        2: "HIGH"
    })
)
```

```python
def trend_risk(trend):
    if trend > 0:
        return 0
    elif trend == 0:
        return 1
    else:
        return 2

df["trend_risk"] = df["performance_trend"].apply(trend_risk)

print("Trend risk distribution:")
print(
    df["trend_risk"]
    .value_counts()
    .sort_index()
    .rename({
        0: "LOW",
        1: "MEDIUM",
        2: "HIGH"
    })
)
```

```python
def failure_risk(failures):
    if failures == 0:
        return 0
    elif failures == 1:
        return 1
    else:
        return 2

df["failure_risk"] = df["failures"].apply(failure_risk)

print("Previous failure risk distribution:")
print(
    df["failure_risk"]
    .value_counts()
    .sort_index()
    .rename({
        0: "LOW",
        1: "MEDIUM",
        2: "HIGH"
    })
)
```

```python
def studytime_risk(studytime):
    if studytime >= 3:
        return 0
    elif studytime == 2:
        return 1
    else:
        return 2

df["studytime_risk"] = df["studytime"].apply(studytime_risk)

print("Study-time risk distribution:")
print(
    df["studytime_risk"]
    .value_counts()
    .sort_index()
    .rename({
        0: "LOW",
        1: "MEDIUM",
        2: "HIGH"
    })
)
```

```python
df["risk_score"] = (
    (df["absence_risk"] / 2) * 25 +
    (df["performance_risk"] / 2) * 30 +
    (df["trend_risk"] / 2) * 20 +
    (df["failure_risk"] / 2) * 15 +
    (df["studytime_risk"] / 2) * 10
)

print("Risk score summary:")
display(df["risk_score"].describe())

print("\nFirst 10 risk scores:")
display(
    df[
        [
            "absences",
            "G1",
            "G2",
            "failures",
            "studytime",
            "performance_trend",
            "risk_score"
        ]
    ].head(10)
)
```

```python
def risk_level(score):
    if score <= 30:
        return "LOW"
    elif score <= 60:
        return "MEDIUM"
    else:
        return "HIGH"


df["risk_level"] = df["risk_score"].apply(risk_level)

print("Risk level distribution:")
print(df["risk_level"].value_counts())

print("\nRisk level percentages:")
print(
    df["risk_level"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

```python
high_risk_students = df[
    df["risk_level"] == "HIGH"
][
    [
        "absences",
        "G1",
        "G2",
        "failures",
        "studytime",
        "performance_trend",
        "absence_risk",
        "performance_risk",
        "trend_risk",
        "failure_risk",
        "studytime_risk",
        "risk_score",
        "risk_level"
    ]
].sort_values(
    "risk_score",
    ascending=False
)

print("High-risk students:", len(high_risk_students))

display(high_risk_students.head(20))
```

```python
def calculate_risk(
    absences,
    g1,
    g2,
    failures,
    studytime
):
    # 1. Absence risk
    if absences <= 4:
        absence_risk = 0
    elif absences <= 9:
        absence_risk = 1
    else:
        absence_risk = 2

    # 2. Current performance risk
    if g2 >= 12:
        performance_risk = 0
    elif g2 >= 10:
        performance_risk = 1
    else:
        performance_risk = 2

    # 3. Performance trend
    performance_trend = g2 - g1

    if performance_trend > 0:
        trend_risk = 0
    elif performance_trend == 0:
        trend_risk = 1
    else:
        trend_risk = 2

    # 4. Previous failure risk
    if failures == 0:
        failure_risk = 0
    elif failures == 1:
        failure_risk = 1
    else:
        failure_risk = 2

    # 5. Study-time risk
    if studytime >= 3:
        studytime_risk = 0
    elif studytime == 2:
        studytime_risk = 1
    else:
        studytime_risk = 2

    # Weighted risk score: 0–100
    risk_score = (
        (absence_risk / 2) * 25 +
        (performance_risk / 2) * 30 +
        (trend_risk / 2) * 20 +
        (failure_risk / 2) * 15 +
        (studytime_risk / 2) * 10
    )

    # Risk level
    if risk_score <= 30:
        risk_level = "LOW"
    elif risk_score <= 60:
        risk_level = "MEDIUM"
    else:
        risk_level = "HIGH"

    return {
        "risk_score": risk_score,
        "risk_level": risk_level,
        "absence_risk": absence_risk,
        "performance_risk": performance_risk,
        "trend_risk": trend_risk,
        "failure_risk": failure_risk,
        "studytime_risk": studytime_risk
    }


print("Risk engine function created successfully.")
```

```python
student = df.iloc[0]

risk_result = calculate_risk(
    absences=student["absences"],
    g1=student["G1"],
    g2=student["G2"],
    failures=student["failures"],
    studytime=student["studytime"]
)

print("Student data:")
print(f"Absences   : {student['absences']}")
print(f"G1         : {student['G1']}")
print(f"G2         : {student['G2']}")
print(f"Failures   : {student['failures']}")
print(f"Study time : {student['studytime']}")

print("\nRisk Engine Result:")
print(f"Risk Score : {risk_result['risk_score']:.2f}")
print(f"Risk Level : {risk_result['risk_level']}")

print("\nOriginal DataFrame Result:")
print(f"Risk Score : {student['risk_score']:.2f}")
print(f"Risk Level : {student['risk_level']}")
```

```python
new_student = {
    "absences": 18,
    "g1": 9,
    "g2": 7,
    "failures": 2,
    "studytime": 1
}

risk_result = calculate_risk(
    absences=new_student["absences"],
    g1=new_student["g1"],
    g2=new_student["g2"],
    failures=new_student["failures"],
    studytime=new_student["studytime"]
)

print("New Student Risk Assessment")
print("----------------------------")
print(f"Absences       : {new_student['absences']}")
print(f"G1             : {new_student['g1']}")
print(f"G2             : {new_student['g2']}")
print(f"Previous Failures: {new_student['failures']}")
print(f"Study Time     : {new_student['studytime']}")

print("\nRisk Result")
print("-----------")
print(f"Risk Score : {risk_result['risk_score']:.2f}/100")
print(f"Risk Level : {risk_result['risk_level']}")
```

```python
def get_risk_indicators(
    absences,
    g1,
    g2,
    failures,
    studytime
):
    indicators = []

    # Attendance / absence
    if absences >= 10:
        indicators.append("High Absence")
    elif absences >= 5:
        indicators.append("Moderate Absence")

    # Current performance
    if g2 < 10:
        indicators.append("Low Current Performance")
    elif g2 < 12:
        indicators.append("Moderate Current Performance")

    # Performance trend
    trend = g2 - g1

    if trend < 0:
        indicators.append("Declining Performance")
    elif trend == 0:
        indicators.append("Stable Performance")

    # Previous failures
    if failures >= 2:
        indicators.append("Multiple Previous Failures")
    elif failures == 1:
        indicators.append("Previous Failure")

    # Study time
    if studytime == 1:
        indicators.append("Low Study Time")
    elif studytime == 2:
        indicators.append("Moderate Study Time")

    return indicators


indicators = get_risk_indicators(
    absences=new_student["absences"],
    g1=new_student["g1"],
    g2=new_student["g2"],
    failures=new_student["failures"],
    studytime=new_student["studytime"]
)

print("Risk Indicators:")
for indicator in indicators:
    print("-", indicator)
```

```python
def assess_student_risk(
    absences,
    g1,
    g2,
    failures,
    studytime
):
    # Calculate score and risk level
    risk_result = calculate_risk(
        absences=absences,
        g1=g1,
        g2=g2,
        failures=failures,
        studytime=studytime
    )

    # Generate actionable indicators
    indicators = get_risk_indicators(
        absences=absences,
        g1=g1,
        g2=g2,
        failures=failures,
        studytime=studytime
    )

    return {
        "risk_score": round(risk_result["risk_score"], 2),
        "risk_level": risk_result["risk_level"],
        "risk_indicators": indicators
    }


print("Final risk assessment function created successfully.")
```

```python
final_assessment = assess_student_risk(
    absences=18,
    g1=9,
    g2=7,
    failures=2,
    studytime=1
)

print("Final Student Risk Assessment")
print("-----------------------------")
print(f"Risk Score : {final_assessment['risk_score']}/100")
print(f"Risk Level : {final_assessment['risk_level']}")

print("\nRisk Indicators:")
for indicator in final_assessment["risk_indicators"]:
    print(f"- {indicator}")
```

```python
risk_results = df.apply(
    lambda row: assess_student_risk(
        absences=row["absences"],
        g1=row["G1"],
        g2=row["G2"],
        failures=row["failures"],
        studytime=row["studytime"]
    ),
    axis=1
)

risk_results_df = pd.DataFrame(risk_results.tolist())

print("Risk assessment completed for all students.")
print("Number of students:", len(risk_results_df))

print("\nRisk level distribution:")
print(risk_results_df["risk_level"].value_counts())

print("\nRisk score summary:")
display(risk_results_df["risk_score"].describe())
```

```python
risk_dataset = df.copy()

risk_dataset["risk_score"] = risk_results_df["risk_score"]
risk_dataset["risk_level"] = risk_results_df["risk_level"]
risk_dataset["risk_indicators"] = risk_results_df["risk_indicators"]

print("Final risk dataset shape:", risk_dataset.shape)

print("\nRisk dataset columns:")
print(risk_dataset.columns.tolist())

print("\nSample risk assessments:")
display(
    risk_dataset[
        [
            "absences",
            "G1",
            "G2",
            "failures",
            "studytime",
            "performance_trend",
            "risk_score",
            "risk_level",
            "risk_indicators"
        ]
    ].head(10)
)
```

```python
student_base = {
    "absences": 8,
    "g1": 10,
    "g2": 9,
    "failures": 1,
    "studytime": 2
}

risk_a = assess_student_risk(**student_base)

# G3 is intentionally changed, but it is NOT passed to the risk engine.
student_with_different_g3 = {
    **student_base,
    "g3": 20
}

risk_b = assess_student_risk(
    absences=student_with_different_g3["absences"],
    g1=student_with_different_g3["g1"],
    g2=student_with_different_g3["g2"],
    failures=student_with_different_g3["failures"],
    studytime=student_with_different_g3["studytime"]
)

print("Risk with G3 = original/unknown:")
print(risk_a)

print("\nRisk with G3 = 20:")
print(risk_b)

print("\nRisk results identical:",
      risk_a == risk_b)
```

```python
risk_config = {
    "weights": {
        "absence": 25,
        "performance": 30,
        "trend": 20,
        "failure": 15,
        "studytime": 10
    },
    "thresholds": {
        "risk_level": {
            "low_max": 30,
            "medium_max": 60
        },
        "absence": {
            "low_max": 4,
            "medium_max": 9
        },
        "performance": {
            "low_min": 12,
            "medium_min": 10
        },
        "failures": {
            "low": 0,
            "medium": 1,
            "high_min": 2
        },
        "studytime": {
            "high": 1,
            "medium": 2,
            "low_min": 3
        }
    }
}

print("Risk engine configuration:")
print(risk_config)
```

```python
test_cases = [
    {
        "name": "Low Risk Student",
        "absences": 0,
        "g1": 15,
        "g2": 16,
        "failures": 0,
        "studytime": 4
    },
    {
        "name": "Medium Risk Student",
        "absences": 5,
        "g1": 11,
        "g2": 10,
        "failures": 1,
        "studytime": 2
    },
    {
        "name": "High Risk Student",
        "absences": 20,
        "g1": 8,
        "g2": 6,
        "failures": 2,
        "studytime": 1
    }
]

for student in test_cases:
    result = assess_student_risk(
        absences=student["absences"],
        g1=student["g1"],
        g2=student["g2"],
        failures=student["failures"],
        studytime=student["studytime"]
    )

    print(student["name"])
    print(f"  Risk Score : {result['risk_score']}/100")
    print(f"  Risk Level : {result['risk_level']}")
    print(f"  Indicators : {result['risk_indicators']}")
    print()
```
